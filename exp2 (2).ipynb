```python
import cv2
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time
```

```python
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Camera not available")
else:
    ret, frame = cap.read()

    if ret:
        cv2.imwrite('captured_frame.jpg', frame)
        print("Image saved successfully")
    else:
        print("Unable to capture image")

cap.release()
```

```python
captured_image = cv2.imread('captured_frame.jpg')

if captured_image is not None:
    captured_image = cv2.cvtColor(captured_image, cv2.COLOR_BGR2RGB)

    plt.imshow(captured_image)
    plt.title('Captured Frame')
    plt.axis('off')
    plt.show()
else:
    print("Captured image not found")
```

```python
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Camera not available")
else:
    for i in range(50):
        ret, frame = cap.read()

        if not ret:
            break

        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

        clear_output(wait=True)
        plt.imshow(frame)
        plt.title('Live Video')
        plt.axis('off')
        plt.show()

        time.sleep(0.05)

cap.release()
```

```python
cap = cv2.VideoCapture(0)

if cap.isOpened():
    for i in range(50):
        ret, frame = cap.read()

        if not ret:
            break

        resized_frame = cv2.resize(frame, (100, 150))
        frame_rgb = cv2.cvtColor(resized_frame, cv2.COLOR_BGR2RGB)

        clear_output(wait=True)
        plt.imshow(frame_rgb)
        plt.title('Resized Video')
        plt.axis('off')
        plt.show()

        time.sleep(0.05)

cap.release()
```

```python
cap = cv2.VideoCapture(0)

if cap.isOpened():
    for i in range(50):
        ret, frame = cap.read()

        if not ret:
            break

        rotated_frame = cv2.rotate(frame, cv2.ROTATE_90_CLOCKWISE)
        frame_rgb = cv2.cvtColor(rotated_frame, cv2.COLOR_BGR2RGB)

        clear_output(wait=True)
        plt.imshow(frame_rgb)
        plt.title('Rotated Video')
        plt.axis('off')
        plt.show()

        time.sleep(0.05)

cap.release()
```
